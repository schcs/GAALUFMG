```python
m1 = Matrix([[2,2,2,0],[-2,5,2,1],[8,1,4,-1]])
m1
```

```python
# Step 1: Eliminate below the first pivot (2)
# R2 = R2 + R1 (to eliminate -2 in position (1,0))
# R3 = R3 - 4*R1 (to eliminate 8 in position (2,0))
m1_step1 = m1.copy()
m1_step1[1,:] = m1_step1[1,:] + m1_step1[0,:]
m1_step1[2,:] = m1_step1[2,:] - 4*m1_step1[0,:]
print("After Step 1:")
m1_step1
```

```python
# Step 2: Eliminate below the second pivot (7)
# R3 = R3 + R2 (to eliminate -7 in position (2,1))
m1_step2 = m1_step1.copy()
m1_step2[2,:] = m1_step2[2,:] + m1_step2[1,:]
print("After Step 2:")
m1_step2
```

```python
# The matrix is now in row echelon form
print("Final row echelon form:")
print(m1_step2)
print("\nInterpretation:")
print("Row 1: 2x + 2y + 2z = 0")
print("Row 2: 7y + 4z = 1") 
print("Row 3: 0 = 0")
print("\nSince the last row gives 0 = 0 (which is always true),")
print("this system has INFINITELY MANY SOLUTIONS.")
print("The system is consistent but has a free variable (z).")
```
