```python
m1 = Matrix([[2,2,2,0],[-2,5,2,1],[8,1,4,-1]])
m1, m1.rref()
```

```python
# Step 1: Eliminate below the first pivot (2)
# R2 = R2 + R1 (to eliminate -2 in position (1,0))
# R3 = R3 - 4*R1 (to eliminate 8 in position (2,0))
m1[1,:] = m1[1,:] + m1[0,:]
m1[2,:] = m1[2,:] - 4*m1[0,:]
m1
```

```python
m1[0,:] = Integer(1)/2*m1[0,:]
m1[1,:] = Integer(1)/7*m1[1,:]
m1
```

```python
m1[0,:] = m1[0,:] - m1[1,:]
m1[2,:] = m1[2,:] + 7*m1[1,:]
m1
```
